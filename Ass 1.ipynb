```python

print("Smart Home Climate & Energy Control")

while True:

    print("\n1. Climate Control")
    print("2. Energy Control")
    print("3. Window Control")
    print("4. Exit")

    op = int(input("Enter your choice: "))

    if op == 1:
        temp = int(input("Enter temperature: "))

        if temp < -40 or temp > 60:
            print("Invalid Temperature")

        elif temp <= 20:
            print("Heater ON")
            if temp <= 10:
                print("Heater Power Increased")

        elif temp < 40:
            print("Fans ON and Windows OPEN")
            if temp >= 35:
                print("Fan Power Increased to 4")

        else:
            print("AC ON, Fans OFF and Windows CLOSED")

        hum = int(input("Enter humidity: "))

        if hum < 0 or hum > 100:
            print("Invalid Humidity")
        elif hum < 50:
            print("Humidifier ON")
        else:
            print("Humidifier OFF")

    elif op == 2:
        tim = int(input("Enter device ON time: "))

        if tim < 0 or tim > 150:
            print("Invalid Time")

        elif tim > 30:
            ans = input("Keep device ON? (Y/N): ").upper()

            if ans == "N":
                print("Device OFF")
            elif ans == "Y":
                print("Device Kept ON")
            else:
                print("Invalid Choice")

        else:
            person = input("Is anyone in the room? (Y/N): ").upper()

            if person == "Y":
                print("Device Kept ON")
            elif person == "N":
                print("Device OFF")
            else:
                print("Invalid Choice")

    elif op == 3:
        rain = input("Is it raining? (Y/N): ").upper()

        if rain == "Y":
            print("Windows CLOSED")
        elif rain == "N":
            print("Windows OPEN")
        else:
            print("Invalid Choice")

    elif op == 4:
        print("Shutting down...")
        break

    else:
        print("Invalid Choice")
```

```python
 
print("================================")
print(" INDUSTRIAL TEMPERATURE CONTROLLER")
print("================================")

while True:

    unit = input("Enter temperature unit (C/F/K): ").upper()

    if unit == "C":
        temp = float(input("Enter temperature: "))
        celsius = temp

    elif unit == "F":
        temp = float(input("Enter temperature: "))
        celsius = (temp - 32) * 5 / 9

    elif unit == "K":
        temp = float(input("Enter temperature: "))
        celsius = temp - 273.15

    else:
        print("Invalid Unit!")
        continue

    fahrenheit = (celsius * 9 / 5) + 32
    kelvin = celsius + 273.15

    print("\nCelsius:", round(celsius, 2), "°C")
    print("Fahrenheit:", round(fahrenheit, 2), "°F")
    print("Kelvin:", round(kelvin, 2), "K")

    if celsius < -50 or celsius > 800:
        print("INVALID TEMPERATURE")

    elif celsius <= -44:
        print("Status: VERY LOW")
        print("Cooling OFF | Heater ON | Buzzer OFF")

    elif celsius > 755:
        print("Status: DANGEROUS")
        print("Cooling ON | Heater OFF | Buzzer ON")

    elif celsius >= 745:
        print("Status: VERY HIGH")
        print("Cooling ON | Heater OFF | Buzzer OFF")

    elif celsius > 400:
        print("Status: HIGH")
        print("Cooling ON | Heater OFF | Buzzer OFF")

    elif celsius < 400:
        print("Status: LOW")
        print("Cooling OFF | Heater ON | Buzzer OFF")

    else:
        print("Status: NORMAL")
        print("Cooling OFF | Heater OFF | Buzzer OFF")
```
