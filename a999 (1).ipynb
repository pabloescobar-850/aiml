```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
mall = pd.read_csv("Mall_Customers.csv")
```

```python
mall.head()
```

```python
mall.info()
```

```python
mall.isnull().sum()
```

```python
mall.duplicated().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

mall["Genre"] = le.fit_transform(mall["Genre"])
```

```python
mall1 = mall[["Annual Income (k$)", "Spending Score (1-100)"]]
```

```python

mall1.head()
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
```

```python
kmeans.fit(mall1)
```

```python
pred = kmeans.predict(mall1)
```

```python
pred
```

```python
kmeans.cluster_centers_
```

```python
kmeans.labels_
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(x=mall["Annual Income (k$)"], y=mall["Spending Score (1-100)"], c=pred)

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("K-Means Clustering (3 Clusters)")
plt.show()
```

```python
wcss = []

for i in range(1, 11):
    km = KMeans(n_clusters=i, random_state=42, n_init=10)
    km.fit(mall1)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1, 11), wcss, color='b')
plt.xticks(range(1, 11))
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

```python
kmeans1 = KMeans(n_clusters=5, random_state=42, n_init=10)
```

```python
kmeans1.fit(mall1)
```

```python
cluster_centers = kmeans1.cluster_centers_
```

```python
kmeans1.labels_
```

```python
plt.scatter(x=mall["Annual Income (k$)"], y=mall["Spending Score (1-100)"], c=kmeans1.labels_)
plt.scatter(cluster_centers[:, 0],cluster_centers[:, 1],marker="o",color="r",label="Cluster Centres")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("K-Means Clustering (5 Clusters)")
plt.legend()
plt.show()
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
z = linkage(mall1, method="single")
```

```python
dendrogram(z)
plt.title("Hierarchical Clustering")
plt.show()
```

```python
z1 = linkage(mall1, method="complete")
```

```python
dendrogram(z1)
plt.title("Hierarchical Clustering - Complete Linkage")
plt.show()
```

```python
z2 = linkage(mall1, method="average")
```

```python
dendrogram(z2)
plt.title("Hierarchical Clustering - Average Linkage")
plt.show()
```
