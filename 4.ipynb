```python
rules = [
    ({"computer not starting", "clicking sound"}, "power problem"),
    ({"power problem"}, "checking power supply"),
    ({"computer on", "no display"}, "display problem"),
    ({"display_problem"}, "check monitor"),
    ({"wifi connected", "no internet"}, "internet_problem"),
    ({"internet problem"}, "restart router")
]
```

```python
solutions = ["checking power supply", "check monitor", "restart router"]
```

```python
def getSymptoms():
    facts = set()

    questions = {
        "computer not starting": "Computer not starting?",
        "clicking sound": "Clicking sound?",
        "computer on": "Computer on?",
        "no display": "No display?",
        "wifi connected": "Wifi connected?",
        "no internet": "No internet?"
    }

    for fact, question in questions.items():
        ans = input(question + " (y/n): ").lower()
        if ans == "y":
            facts.add(fact)
    return facts
```

```python
def forward(facts):
    changed = True
    while changed:
        changed = False
        for conditions, result in rules:
            if conditions.issubset(facts) and result not in facts:
                facts.add(result)
                print("Found", result)
                changed = True
    print("\n Solution")
    for s in solutions:
        if s in facts:
            print(s)
```

```python
def backward(goal, facts):
    if goal in facts:
        return True
    for conditions , result in rules:
        if result == goal:
            return all(backward (c, facts) for c in conditions)

    return False
```

```python
while True:
    print("\n1)forward chaning")
    print("2)backward chaning")
    print("3)EXIT")

    choice = (input("Enter teh choice :?"))
    if choice == "1":
        forward(getSymptoms())
    elif choice == "2":
        facts = getSymptoms()
        goal = input("Enter Goal")
        print("Goal proved" if backward(goal, facts) else "Goal Not Found" )

    elif choice == "3":
        print("EXITING")
        break
```
