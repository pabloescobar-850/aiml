```python
import heapq

def gbfs(graph, heuristic, start, end):
    open_list = []
    heapq.heappush(open_list, (heuristic[start], start))

    closed_list = set()
    ans = []


    while open_list:
        h, current = heapq.heappop(open_list)

        if current in closed_list:
            continue

        if current == end:
            print("Final Path:", ans)
            return

        for neigh in graph[current]:
            if neigh not in closed_list:
                heapq.heappush(
                    open_list,
                    (heuristic[neigh], neigh)
                )
    print("Path Not Found")


n = int(input("Enter number of nodes:"))

graph = {}

for i in range(n):
    node = input("Enter The Nodes:")
    num_neighbours = int(input("Enter the Number of Neighbours:"))

    neighbours = []

    for j in range(num_neighbours):
        nei = input("Enter the Neighbours: ")
        neighbours.append(nei)

    graph[node] = neighbours

print("Graph is :", graph)
heuristic = {}
print("Enter the Heuristic values :")

for node in graph:
    heuristic[node] = int(input("Emter the value of " + node + ":"))

start = int(input("Entet where to start:"))
end = input("Enter where to end :")

gbfs(graph, heuristic, start, end)
```
