```python

from collections import deque

def bfs(graph, start, goal):
    visited = {start}
    queue = deque([start])

    while queue:
        node = queue.popleft()
        print(node)

        if node == goal:
            print("Goal Found")
            return

        for n in graph[node]:
            if n not in visited:
                visited.add(n)
                queue.append(n)


def dfs(graph, start, goal):
    visited = {start}
    stack = [start]

    while stack:
        node = stack.pop()
        print(node)

        if node == goal:
            print("Goal Found")
            return

        for n in graph[node]:
            if n not in visited:
                visited.add(n)
                stack.append(n)


graph = {}
count = int(input())

for i in range(count):
    node = input("Enter node: ")
    graph[node] = []

    total = int(input("Enter number of connected nodes: "))

    for j in range(total):
        graph[node].append(input("Enter connected node: "))


print("Graph:", graph)

start = input("Enter start node: ")
goal = input("Enter goal node: ")

while True:
    print("\n1. BFS")
    print("2. DFS")
    print("3. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        bfs(graph, start, goal)

    elif choice == 2:
        dfs(graph, start, goal)

    elif choice == 3:
        print("Exiting...")
        break

    else:
        print("Invalid Choice")
```
